<h2><b>2.1 Landasan Teori</b></h2>

**​2.1.1 Konsep Dasar Control Flow dan Seleksi**

​Struktur kontrol alur (_control flow_) merupakan mekanisme dalam bahasa pemrograman yang mengatur urutan eksekusi instruksi atau baris kode oleh _interpreter_. Dalam eksekusi standar, program berjalan secara sekuensial dari atas ke bawah. Namun, untuk membangun program yang dinamis, diperlukan pemrosesan seleksi (_selection_) yang memungkinkan program memilih jalur eksekusi tertentu berdasarkan evaluasi ekspresi kondisi yang bernilai boolean (`True` atau `False`).

**​2.1.2 Jenis-Jenis Struktur Kondisional pada Python**

**​1. Percabangan Tunggal (`if`)**

​Struktur `if` merupakan bentuk percabangan paling dasar. Blok kode di dalam pernyataan `if` hanya akan dieksekusi apabila ekspresi kondisi yang diuji mengevaluasi nilai `True`. Jika kondisi bernilai `False`, _interpreter_ akan melompati blok kode tersebut.

**​2. Percabangan Ganda (`if-else`)**

​Struktur `if-else` menyediakan alur alternatif dua arah. Apabila kondisi pada pernyataan `if` bernilai `True`, maka blok kode `if` yang dieksekusi. Sebaliknya, jika kondisi bernilai `False`, eksekusi program dialihkan secara otomatis ke blok kode di bawah klausul `else`.

**​3. Percabangan Majemuk (`if-elif-else`)**

​Untuk mengevaluasi lebih dari dua kemungkinan kondisi secara berurutan, Python menyediakan kata kunci `elif` (singkatan dari `else if`). _Interpreter_ akan menguji kondisi dari atas ke bawah secara hirarkis dan mengeksekusi blok kode pertama yang bernilai `True`. Jika seluruh rantai kondisi `if` dan `elif` bernilai `False`, blok `else` terakhir yang akan dieksekusi sebagai jalur _default_.

**​4. Percabangan Bersarang (Nested IF)**

​Percabangan bersarang merupakan kondisi di mana sebuah pernyataan `if`, `if-else`, atau `if-elif-else` diletakkan di dalam blok kode pernyataan `if` lainnya. Struktur ini digunakan untuk memproses pengujian kondisi bertingkat yang memiliki ketergantungan logika secara hierarkis.

**​5. Ekspresi Kondisional (Ternary Operator)**

​Python mendukung penulisan pernyataan kondisional sederhana dalam satu baris kode (_one-liner_) menggunakan ekspresi _ternary_. Sintaks ini digunakan untuk mempersingkat pengalokasian nilai variabel berdasarkan kondisi tertentu dengan bentuk umum:

`nilai_jika_true if kondisi else nilai_jika_false`.

**6. Pencocokan Pola Terstruktur (`match-case`)**

​Mulai dari Python versi 3.10, Python memperkenalkan fitur _Structural Pattern Matching_ menggunakan kata kunci `match` dan `case`. Struktur ini mirip dengan pernyataan `switch-case` pada bahasa pemrograman lain seperti C++ atau Java, tetapi memiliki kemampuan pencocokan pola (_pattern matching_) yang lebih fleksibel.

​Struktur `match-case` mengevaluasi suatu nilai atau variabel terhadap beberapa pola (_case_) yang didefinisikan. Karakter `_` (_wildcard_) digunakan pada `case _:` sebagai jalur alternatif (_default case_) yang akan dieksekusi apabila tidak ada pola di atasnya yang cocok.

```python
match variabel:
    case nilai_1:
        # blok kode jika variabel == nilai_1
    case nilai_2:
        # blok kode jika variabel == nilai_2
    case _:
        # blok kode default jika tidak ada pola yang cocok
```

<h3><b>2.2.1 Program "if_elif_else1_2611532001.py"</b></h3>

```python
# Buat program untuk kondisional if
# Nama variabel ditambah 4 digit nim terakhir contoh: ipk_1234
# Program ini menggunakan fungsi input()

umur_2001 = int(input("Input Umur Anda: "))
sim_2001 = input("Apakah Anda Sudah Punya Sim C (y/t): ")[0]

if umur_2001 >= 17 and sim_2001 == 'y':
    print("Anda Sudah Dewasa dan Boleh Bawa Motor")
elif umur_2001 >= 17 and sim_2001 != 'y':
    print("Anda Sudah Dewasa tetapi Tidak Boleh Bawa Motor")
elif umur_2001 < 17 and sim_2001 == 'y':
    print("Anda Belum Cukup Umur Punya SIM")
else:
    print("Anda Belum Cukup Umur Bawa Motor")

print("Program Selesai")
```

**Penjelasan :**

**1. Membuat Komentar dan Menerima Input Data:** Baris diawali tanda `#` berfungsi sebagai komentar penjelas. Sintaks `input()` mengambil masukan umur dari pengguna yang dikonversi ke tipe integer (`int()`) pada variabel `umur_2001`. Sintaks `input()[0]` pada variabel `sim_2001` mengambil karakter pertama dari string masukan untuk memastikan pengujian karakter (`'y'` atau `'t'`) berjalan secara tepat.

**2. ​Mengevaluasi Kondisi Utama (`if`):** Sintaks `if umur_2001 >= 17 and sim_2001 == 'y':` menggabungkan dua kriteria dengan operator logika `and`. Blok `print()` di dalamnya hanya dieksekusi jika pengguna berusia minimal 17 tahun dan telah memiliki SIM C (`True`).

**3. ​Mengevaluasi Kondisi Alternatif Pertama (`elif` ke-1):** Sintaks `elif umur_2001 >= 17 and sim_2001 != 'y':` mengevaluasi kasus di mana pengguna sudah genap/lebih dari 17 tahun namun belum memiliki SIM C (`sim_2001 != 'y'`), sehingga mencetak pesan bahwa pengguna belum diperbolehkan mengendarai motor.

**4. ​Mengevaluasi Kondisi Alternatif Kedua (`elif` ke-2):** Sintaks `elif umur_2001 < 17 and sim_2001 == 'y':` mengevaluasi kasus kontradiktif di mana umur pengguna kurang dari 17 tahun tetapi mengklaim memiliki SIM C, sehingga memberikan peringatan bahwa usia tersebut belum cukup umur untuk penerbitan SIM.

**6. ​Menangani Kondisi Default (`else`) dan Penutup Program:** Klausul `else`: menangani seluruh kondisi lainnya (yaitu ketika umur di bawah 17 tahun dan tidak memiliki SIM C). Perintah `print("Program Selesai")` diletakkan di luar blok percabangan (tanpa indentasi) sehingga akan selalu dieksekusi sebagai penanda bahwa alur eksekusi program telah berakhir.

<h3><b>2.2.4 Program "match_case1_2611532001.py"</b></h3>

```python
# Buat program untuk nested if
# Nama variabel ditambah 4 digit nim terakhir contoh: ipk_1234
# Program ini menggunakan fungsi input()
# Program konversi nilai angka menjadi nama bulan

bulan_2001 = int(input("Masukkan Angka Bulan (1 - 12): "))

match bulan_2001:
    case 1:
        print("Januari")
    case 2:
        print("Februari")
    case 3:
        print("Maret")
    case 4:
        print("April")
    case 5:
        print("Mei")
    case 6:
        print("Juni")
    case 7:
        print("Juli")
    case 8:
        print("Agustus")
    case 9:
        print("September")
    case 10:
        print("Oktober")
    case 11:
        print("November")
    case 12:
        print("Desember")
    case _:
        print("Angka Tidak Valid")
```

**Penjelasan :**

**1. Membuat Komentar dan Menerima Input Angka:** Tanda `#` digunakan sebagai komentar penjelas. Sintaks `input()` menangkap masukan angka dari pengguna secara interaktif, yang kemudian dikonversi secara eksplisit menggunakan fungsi `int()` menjadi tipe data bilangan bulat (integer) pada variabel `bulan_2001`.

**2. ​Menerapkan Struktur Pencocokan Pola (`match-case`):** Sintaks `match bulan_2001:` mengevaluasi nilai variabel terhadap serangkaian pola alternatif menggunakan kata kunci `case`. Struktur ini bertindak sebagai percabangan majemuk modern yang memeriksa nilai dari angka 1 hingga 12 secara berurutan.

**3. ​Mencetak Nama Bulan:** Masing-masing blok `case` (mulai dari `case 1:` hingga `case 12:`) bertugas mencetak nama bulan yang sesuai (seperti `"Januari"`, `"Februari"`, dst.) ke layar konsol apabila nilai variabel cocok dengan angka kasus tersebut.

**4. ​Menangani Kondisi Alternatif (Wildcard `_`): Klausul `case _:` berfungsi sebagai jalur default atau penanganan galat (_error handling_). Karakter garis bawah `_` bertindak sebagai _wildcard_ yang akan menangkap seluruh masukan angka di luar rentang 1 hingga 12 (misalnya angka negatif atau lebih dari 12), lalu mencetak pesan `"Angka Tidak Valid"`.

<h3><b>2.2.5 Program "multi_if1_2611532001.py"</b></h3>

```python
# Buat program untuk kondisional if
# Nama variabel ditambah 4 digit nim terakhir contoh: ipk_1234
# Program ini menggunakan fungsi input()

umur_2001 = int(input("Input Umur Anda: "))
sim_2001 = input("Apakah Anda Sudah Punya Sim C (y/t): ")[0]

if umur_2001 >= 17 and sim_2001 == 'y':
    print("Anda Sudah Dewasa dan Boleh Bawa Motor")

if umur_2001 >= 17 and sim_2001 != 'y':
    print("Anda Sudah Dewasa tetapi Tidak Boleh Bawa Motor")

if umur_2001 < 17 and sim_2001 == 'y':
    print("Anda Belum Cukup Umur Punya SIM")

if umur_2001 < 17 and sim_2001 != 't':
    print("Anda Belum Cukup Umur Bawa Motor")

print("Program Selesai")
```

**4. ​Mengevaluasi Blok `if` Keempat dengan Logika Negasi:** Pernyataan `if umur_2001 < 17 and sim_2001 != 't':` memeriksa kategori usia di bawah 17 tahun yang memasukkan status selain huruf `'t'` (tidak), untuk memastikan pembatasan hak berkendara bagi anak di bawah umur.

Baris `print("Program Selesai")` diletakkan di luar seluruh blok struktur kondisional sehingga akan selalu dieksekusi di akhir alur program.

<h3><b>2.2.6 Program "multi_if2_2611532001.py"</b></h3>

```python
# Buat program untuk kondisional if
# Nama variabel ditambah 4 digit nim terakhir contoh: ipk_1234
# Program ini menggunakan fungsi input()
# Program menghitung diskon belanja

# Input dari user
total_belanja_2001 = float(input("Masukkan Total Belanja (Rp): "))

# Input status member (mengecek apakah user mengetik 'y' atau 'ya')
input_member_2001 = input("Apakah Anda Member? (y/t): ").strip().lower()
is_member_2001 = input_member_2001 in ["y", "ya"]

# Input status kode promo (mengecek apakah user mengetik 'y' atau 'ya')
input_promo_2001 = input("Apakah Kode Promo Valid (y/t): ").strip().lower()
kode_promo_valid_2001 = input_promo_2001 in ["y", "ya"]

total_diskon_persen_2001 = 0

# Multi-IF terpisah: Setiap kondisi diperiksa secara independen
# Diskon bisa ditumpuk (akumulasi) jika memenuhi beberapa syarat sekaligus
if total_belanja_2001 > 1000000:
    total_diskon_persen_2001 += 10 # Diskon belanja besar

if is_member_2001:
    total_diskon_persen_2001 += 5 # Diskon belanja member

if kode_promo_valid_2001:
    total_diskon_persen_2001 += 15 # Diskon belanja voucher

# Menghitung nominal diskon dan total bayar
nominal_diskon_2001 = total_belanja_2001 * (total_diskon_persen_2001 / 100)
total_bayar_2001 = total_belanja_2001 - nominal_diskon_2001

# Output hasil
print("\n--- Rincian Pembayaran ---")
print(f"Total Diskon : {total_diskon_persen_2001}% (Rp {nominal_diskon_2001:,.0f})")
print(f"Total Bayar  : Rp {total_bayar_2001:,.0f}")

print(f"Total Diskon yang Anda Dapatkan: {total_diskon_persen_2001}%")
# Output: Total diskon yang anda dapatkan: 30% jika belanja > 1 juta, member, dan kode promo valid
```

**Penjelasan:**

**1. Membuat Komentar dan Menerima Input Data Multi-Tipe:** Tanda `#` digunakan sebagai komentar penjelas. Sintaks `input()` menangkap nilai nominal total belanja yang dikonversi ke tipe data pecahan (`float()`) pada variabel `total_belanja_2001`. Masukan status keanggotaan dan kupon promo dibersihkan menggunakan metode `.strip().lower()` untuk memastikan validasi teks tahan terhadap spasi atau perbedaan huruf besar-kecil.

Perintah pencetakan menggunakan f-string dengan pemformatan koma desimal (`:,.0f`) untuk menampilkan nominal angka rupiah ke layar konsol secara rapi dan profesional.

```python
# Buat program untuk nested if
# Nama variabel ditambah 4 digit nim terakhir contoh: ipk_1234
# Program ini menggunakan fungsi input()
# Program konversi nilai angka menjadi huruf

nilai_2001 = int(input("Inputkan Nilai Angka = "))

if nilai_2001 >= 81:
    print("A")
elif nilai_2001 >= 70:
    print("B")
elif nilai_2001 >= 60:
    print("C")
elif nilai_2001 >= 50:
    print("D")
else :
    print("E")
```

**Penjelasan:**

**1. Membuat Komentar dan Menerima Input Angka:** Tanda `#` digunakan sebagai komentar penjelas. Sintaks `input()` menangkap masukan nilai angka dari pengguna secara interaktif, yang kemudian dikonversi secara eksplisit (_type casting_) menggunakan fungsi `int()` menjadi bilangan bulat (integer) pada variabel `nilai_2001`.

**2. ​Mengevaluasi Kondisi Ambang Batas Tertinggi (`if`):** Sintaks `if nilai_2001 >= 81:` bertindak sebagai gerbang pengujian pertama. Jika nilai angka yang dimasukkan praktikan bernilai 81 atau lebih, program langsung mengeksekusi perintah `print("A")` dan melompati seluruh sisa cabang kondisi di bawahnya.

**3. ​Menerapkan Percabangan Bertingkat Majemuk (elif):** Rangkaian klausul `elif nilai_2001 >= 70:`, `elif nilai_2001 >= 60:`, dan `elif nilai_2001 >= 50:` mengevaluasi rentang ambang batas nilai secara berurutan dari atas ke bawah. Sifat hirarkis ini memastikan bahwa interpreter hanya akan mengeksekusi blok kode pertama yang memenuhi syarat, sementara nilai yang lebih rendah akan disaring secara otomatis oleh batas kondisi sebelumnya.

**4. ​Menangani Kasus Alternatif Terakhir (`else`):** Klausul `else:` berfungsi sebagai jalur default penutup apabila seluruh pengujian kondisi di atas bernilai `False` (yaitu ketika nilai angka kurang dari 50). Blok ini akan mencetak predikat huruf `"E"` ke layar konsol.

```python
# ==========================================
# 1. HEADER DAN INPUT DATA PENGUNJUNG
# ==========================================
print("=== SISTEM LOKET ALPRO ADVENTURE PARK ===")
nama_2001 = input("Masukkan Nama Pengunjung     : ")
umur_2001 = int(input("Input Umur Anda              : "))
is_simC_2001 = input("Apakah Anda Punya SIM C (y/t): ").strip().lower()[0]

print("\nPilihan Paket Wahana (1-5)")
print(" 1. Safari Rimba         (Rp 50,000)")
print(" 2. Arung Jeram          (Rp 75,000)")
print(" 3. Motor ATV Ekstrim    (Rp 120,000)")
print(" 4. Roller Coaster Kilat (Rp 100,000)")
print(" 5. All-Access VIP       (Rp 220,000)")


# ==========================================
# 2. PEMILIHAN PAKET WAHANA (MATCH-CASE)
# ==========================================
while True:
    input_tiket_2001 = int(input("Masukkan nomor paket (1-5)     : "))

    match input_tiket_2001:
        case 1:
            jenis_wahana_2001 = "Wahana Safari Rimba"
            harga_2001 = 50000
            break
        case 2:
            jenis_wahana_2001 = "Wahana Arung Jeram"
            harga_2001 = 75000
            break
        case 3:
            jenis_wahana_2001 = "Wahana Motor ATV Ekstrim"
            harga_2001 = 120000
            break
        case 4:
            jenis_wahana_2001 = "Wahana Roller Coaster Kilat"
            harga_2001 = 100000
            break
        case 5:
            jenis_wahana_2001 = "Wahana All-Access VIP"
            harga_2001 = 220000
            break
        case _:
            print("\nNomor Paket Tidak Valid. Silakan Pilih Kembali!")


# ==========================================
# 3. INPUT JUMLAH TIKET DAN STATUS TAMBAHAN
# ==========================================
while True:
    jumlah_tiket_2001 = int(input("Masukkan jumlah tiket          : "))
    if (jumlah_tiket_2001 > 0):
        break
    else:
        print("\nJumlah Tiket Tidak Valid. Silakan Masukan Kembali!")

is_member_2001 = input("Apakah Anda member? (y/t)      : ").strip().lower()[0]
is_kode_2001 = input("Apakah kode promo valid? (y/t) : ").strip().lower()[0]

is_kondisi_2001 = True
total_diskon_2001 = 0


# ==========================================
# 4. VALIDASI KELAYAKAN PENGENDARA WAHANA
# ==========================================
print("\n--- KELAYAKAN PENGENDARA WAHANA ---")
if (input_tiket_2001 == 3):
    if (umur_2001 >= 17 and is_simC_2001 == "y"):
        print(f"STATUS AKSES : Anda Sudah Dewasa dan Boleh Menaiki {jenis_wahana_2001} Sendiri.")
    elif (umur_2001 >= 17 and is_simC_2001 != "y"):
        print(f"STATUS AKSES : Anda Sudah Dewasa tetapi Tidak Boleh Menaiki {jenis_wahana_2001} Sendiri (Wajib Didampingi Instruktur).")
    elif (umur_2001 < 17 and is_simC_2001 == "y"):
        print(f"IDENTITAS TIDAK VALID: Belum Cukup Umur Memiliki SIM.")
        is_kondisi_2001 = False
    else:
        print(f"Anda Belum Cukup Umur dan Tidak Boleh Menaiki {jenis_wahana_2001}.")
        is_kondisi_2001 = False
else:
    if (umur_2001 >= 10):
        print(f"STATUS AKSES : Batas Umur Tercukupi dan Anda Boleh Menaiki {jenis_wahana_2001}.")
    else:
        print(f"STATUS AKSES : Anda Tidak Cukup Umur dan Tidak Boleh Menaiki {jenis_wahana_2001}.")
        is_kondisi_2001 = False


# ==========================================
# 5. AKUMULASI DISKON & RINCIAN PEMBAYARAN
# ==========================================
if is_kondisi_2001:
    if (harga_2001 * jumlah_tiket_2001 >= 200000):
        total_diskon_2001 += 10
    if (is_member_2001 == "y"):
        total_diskon_2001 += 5
    if (is_kode_2001 == "y"):
        total_diskon_2001 += 15
    if (jumlah_tiket_2001 >= 5):
        total_diskon_2001 += 5

    print("\n--- Rincian Pembayaran ---")
    print(f"Subtotal Belanja : Rp {harga_2001 * jumlah_tiket_2001:,.0f}")
    print(f"Total Diskon     : {total_diskon_2001}% (Rp {(harga_2001 * jumlah_tiket_2001) * (total_diskon_2001 / 100):,.0f})")
    print(f"Total Bayar       : Rp {(harga_2001 * jumlah_tiket_2001) - ((harga_2001 * jumlah_tiket_2001) * (total_diskon_2001 / 100)):,.0f}")
    if ((harga_2001 * jumlah_tiket_2001) - ((harga_2001 * jumlah_tiket_2001) * (total_diskon_2001 / 100)) > 300000):
            print("Selamat! Anda berhak mendapatkan Souvenir Gratis.")
    print("Catatan Layanan  : Terima kasih telah berkunjung.")


# ==========================================
# 6. PENUTUP PROGRAM
# ==========================================
print("\nProgram Selesai")
```
