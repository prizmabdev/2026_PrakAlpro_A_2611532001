<h2><b>2.1 Landasan Teori</b></h2>

​Operator merupakan simbol atau kata kunci khusus yang digunakan untuk mengolah nilai atau variabel (operand) guna menghasilkan nilai baru. Dalam bahasa pemrograman Python, operator dikelompokkan menjadi tujuh kategori utama berdasarkan fungsi dan penggunaannya:

**2.1.​1. Operator Aritmetika (Arithmetic Operators)**

​Operator aritmetika digunakan untuk melakukan pemrosesan matematis atau perhitungan angka dasar. Python mendukung operator penjumlahan (`+`), pengurangan (`-`), perkalian (`*`), pembagian (`/`), pembagian bulat (`//`), sisa bagi atau modulus (`%`), serta operasi perpangkatan (`**`).

**2.1.​2. Operator Penugasan (Assignment Operators)**

​Operator penugasan berfungsi untuk memasukkan atau mengalokasikan nilai ke dalam suatu variabel. Selain operator penugasan dasar (`=`), Python juga menyediakan operator penugasan gabungan (seperti `+=`, `-=`, `*=`, `/=`, dan `%=`) yang berguna untuk menyederhanakan penulisan operasi aritmetika yang hasilnya disimpan kembali ke variabel yang sama.

**2.1.​3. Operator Perbandingan (Relational/Comparison Operators)**

​Operator perbandingan digunakan untuk membandingkan dua nilai atau variabel. Hasil dari operasi perbandingan ini selalu berupa nilai boolean (`True` atau `False`). Operator yang termasuk dalam kelompok ini meliputi sama dengan (`==`), tidak sama dengan (`!=`), lebih besar dari (`>`), lebih kecil dari (`<`), lebih besar dari atau sama dengan (`>=`), serta lebih kecil dari atau sama dengan (`<=`).

**​4. Operator Logika (Logical Operators)**

​Operator logika digunakan untuk menghubungkan satu atau beberapa ekspresi kondisi relasional untuk menghasilkan satu nilai kebenaran akhir. Terdapat tiga operator logika utama pada Python, yaitu:

> * **​and:** Bernilai True hanya jika seluruh kondisi bernilai benar.

> * **​or:** Bernilai True jika salah satu dari kondisi bernilai benar.

> * **​not:** Berfungsi untuk membalikkan nilai logika (mengubah True menjadi False dan sebaliknya).

**​5. Operator Bitwise (Bitwise Operators)**

​Operator bitwise digunakan untuk memanipulasi data angka pada tingkat bit (biner 0 dan 1). Operator ini bekerja dengan mengonversi angka menjadi deretan bit terlebih dahulu. Jenis operator bitwise di Python antara lain AND (`&`), OR (`|`), XOR (`^`), NOT (`~`), Left Shift (`<<`), dan Right Shift (`>>`).

**​6. Operator Keanggotaan (Membership Operators)**

​Operator keanggotaan digunakan untuk menguji keberadaan atau ketersediaan suatu nilai/elemen di dalam sebuah koleksi data beruntun (seperti string, list, tuple, atau dictionary). Operator ini terdiri dari `in` (bernilai True jika elemen ditemukan) dan `not in` (bernilai True jika elemen tidak ditemukan).

**​7. Operator Identitas (Identity Operators)**

​Operator identitas berfungsi untuk membandingkan alokasi alamat memori dari dua buah objek atau variabel. Operator `is` akan bernilai True jika kedua variabel merujuk pada objek memori yang persis sama, sedangkan operator `is not` akan bernilai True jika kedua variabel merujuk pada lokasi memori yang berbeda.

<h3><b>2.2.1 Program "aritmatika_2611532001.py"</b></h3>

```python
# Buat file dengan nama aritmatika_NIM.py
# Buat program untuk operator aritmatika dalam Python
# Nama variabel ditambah 4 digit nim terakhir contoh: angka1_1234
# Program ini menggunakan fungsi input()
# Nilai yang dimasukkan akan dikonversi menjadi tipe data integer

angka1_2001 = int(input("Input angka-1: "))
angka2_2001 = int(input("Input angka-2: "))

# Penjumlahan
hasil_2001 = angka1_2001 + angka2_2001
print("\nOperator Penjumlahan")
print("Hasil =", hasil_2001)

# Pengurangan
hasil_2001 = angka1_2001 - angka2_2001
print("\nOperator Pengurangan")
print("Hasil =", hasil_2001)

# Perkalian
hasil_2001 = angka1_2001 * angka2_2001
print("\nOperator Perkalian")
print("Hasil =", hasil_2001)

# Pembagian, pembagian bulat, dan sisa bagi
if angka2_2001 != 10:
    hasil_2001 = angka1_2001 / angka2_2001
    print("\noperator Pembagian")
    print("Hasil =", hasil_2001)

    hasil_2001 = angka1_2001 // angka2_2001
    print("\nOperator Pembagian Bulat")
    print("Hasil =", hasil_2001)

    hasil_2001 = angka1_2001 % angka2_2001
    print("\nOperator Sisa Bagi")
    print("Hasil =", hasil_2001)
else:
    print("Angka kedua tidak boleh bernilai 0.")

# Pangkat
hasil_2001 = angka1_2001 ** angka2_2001
print("\nOperator Pangkat")
print("Hasil =", hasil_2001)
```

**Penjelasan :**

**1. Membuat Komentar:** Sintaks diawali tanda `#` (seperti `# Buat program untuk operator aritmatika dalam Python`) digunakan sebagai komentar penjelas untuk memberikan keterangan fungsi program, aturan penamaan variabel dengan NIM, serta petunjuk instruksi konversi tipe data.

**2. ​Menerapkan Input Data dan Konversi Tipe Data (Type Casting):** Sintaks `angka1_2001 = int(input("Input angka-1: "))` dan `angka2_2001 = int(input("Input angka-2: "))` digunakan untuk menerima masukan interaktif berupa angka dari pengguna. Fungsi `int()` mengonversi data masukan (yang secara bawaan bertipe string) menjadi bilangan bulat (integer) agar dapat diolah dalam operasi aritmetika.

**3. ​Melakukan Operasi Aritmetika Dasar (Penjumlahan, Pengurangan, Perkalian, Pangkat):** Sintaks `+`, `-`, `*`, dan `**` digunakan untuk memproses kalkulasi matematika. Nilai dari `angka1_2001` dan `angka2_2001` dijumlahkan, dikurangi, dikalikan, serta dipangkatkan, lalu hasilnya disimpan secara bergantian ke dalam variabel `hasil_2001`.

**4. ​Pengondisian Evaluasi Operasi Pembagian, Pembagian Bulat, dan Sisa Bagi:** Sintaks `if angka2_2001 != 10:` memvalidasi nilai penyebut dengan operator perbandingan tidak sama dengan (`!=`). Jika kondisi terpenuhi, program mengeksekusi pembagian desimal dengan operator `/`, pembagian bulat (floor division) dengan `//`, serta sisa bagi (modulus) dengan `%`. Jika tidak, bagian `else` mengeksekusi pesan peringatan.

**5. ​Mengeluarkan Output Program:** Perintah `print()` digunakan untuk menampilkan judul masing-masing jenis operator aritmetika beserta nilai hasil kalkulasi yang tersimpan pada variabel `hasil_2001` ke layar output konsol.

<h3><b>2.2.2 Program "assignment_2611532001.py"</b></h3>

```python
# Buat file dengan nama assignment_NIM.py
# Nama variabel ditambah 4 digit nim terakhir contoh: angkal_1234
# Program ini menggunakan fungsi input()
# Nilai yang dimasukkan akan dikonversi menjadi tipe data Integer
# Program operator assignment dalam Python

angka1_2001 = int(input("Input angka-1: "))
angka2_2001 = int(input("Input angka-2: "))

print("\nNilai awal angka1 =", angka1_2001)
print("Nilai angka2 =", angka2_2001)

# Assignment biasa
hasil_2001 = angka1_2001
print("\nAssignment biasa (=)")
print("Hasil =", hasil_2001)

# Assignment penambahan
hasil_2001 = angka1_2001
hasil_2001 += angka2_2001
print("\nAssignment penambahan (+-)")
print("Hasil =", hasil_2001)

# Assignment pengurangan
hasil_2001 = angka1_2001
hasil_2001 -= angka2_2001
print("\nAssignment pengurangan (-)")
print("Hasil =", hasil_2001)

# Assignment perkalian
angka1_2001 = hasil_2001
angka2_2001 *= hasil_2001
print("\nAssignment perkalian (*=)")
print("Hasil =", hasil_2001)

# Assignment pembagian, pembagian bulat, dan sisa bagi
if angka2_2001 != 0:
    hasil_2001 = angka1_2001
    hasil_2001 /= angka2_2001
    print("\nAssignment pembagian (/-)")
    print("Hasil =", hasil_2001)
    #Operator tambahan
    hasil_2001 = angka1_2001
    hasil_2001 //= angka2_2001
    print("\nAssignment pembagian bulat (//=)")
    print("Hasil =", hasil_2001)
    hasil_2001 = angka1_2001
    hasil_2001 %= angka2_2001
    print("\nAssignment sisa bagi (%=)")
    print("Hasil =", hasil_2001)
else:
    print("\nPembagian tidak dapat dilakukan.")
    print("Angka kedua tidak boleh bernilai 0.")

#Operator tambahan: assignment perpangkatan
hasil_2001 = angka1_2001
hasil_2001 **= angka2_2001
print("\nAssignment perpangkatan (**)")
print("Hasil =", hasil_2001)
```

**Penjelasan :**

**1. Membuat Komentar:** Sintaks diawali tanda `#` (seperti `# Program operator assignment dalam Python`) digunakan sebagai komentar penjelas yang memberikan petunjuk instruksi penamaan variabel berbasis NIM, fungsi `input()`, konversi tipe data ke integer, serta keterangan alur tiap baris kode.

**2. ​Menerapkan Input Data dan Konversi Tipe Data (Type Casting):** Sintaks `angka1_2001 = int(input("Input angka-1: "))` dan `angka2_2001 = int(input("Input angka-2: "))` digunakan untuk menerima masukan interaktif berupa angka dari pengguna. Fungsi `int()` digunakan untuk mengonversi tipe data bawaan string menjadi bilangan bulat (integer).

**3. ​Menerapkan Operator Assignment Dasar dan Gabungan (Penjumlahan, Pengurangan, Perkalian):**

> * ​Sintaks `hasil_2001 = angka1_2001` menggunakan operator assignment dasar (`=`) untuk memasukkan nilai variabel.

> * ​Sintaks `+=`, `-=`, dan `*=` merupakan operator assignment gabungan. Contohnya `hasil_2001 += angka2_2001` secara ringkas melakukan penjumlahan `hasil_2001 = hasil_2001 + angka2_2001` lalu menyimpan kembali nilainya ke variabel `hasil_2001`.

**4. ​Pengondisian dan Operator Assignment Pembagian, Pembagian Bulat, serta Sisa Bagi:** Sintaks `if angka2_2001 != 0:` memeriksa agar penyebut tidak bernilai `0` dengan operator perbandingan. Jika terpenuhi, program mengeksekusi operator assignment pembagian (`/=`), pembagian bulat (`//=`), dan sisa bagi (`%=`). Jika bernilai `0`, perintah pada blok `else` akan dicetak ke layar.

**5. ​Menerapkan Operator Assignment Perpangkatan dan Output Program:** Sintaks `hasil_2001 **= angka2_2001` melakukan operasi perpangkatan dan memperbarui nilai variabel secara langsung. Perintah `print()` di setiap bagian digunakan untuk mencetak label jenis operator assignment beserta nilai akhir variabel `hasil_2001` ke konsol output.

<h3><b>2.2.3 Program "bitwise_2611532001.py"</b></h3>

```python
# Buat file dengan nama bitwise_NIM.py
# Nama variabel ditambah 4 digit nim terakhir contoh: angkal 1234
# Program ini menggunakan fungsi input()

print("\n===================")
print("3. OPERATOR BITWISE")
print("===================")

angka1_2001 = int(input("Masukkan angka bitwise-1: "))
angka2_2001 = int(input("Masukkan angka bitwise-2: "))

print("\nAngka dalam bentuk desimal dan biner")
print("angka1_2001 =", angka1_2001, "| biner", bin(angka1_2001))
print("angka2_2001 =", angka2_2001, "| biner", bin(angka2_2001))

# Bitwise AND
hasil_2001 = angka1_2001 & angka2_2001
print("\nBitwise AND (&)")
print(angka1_2001, "&", angka2_2001, "=", hasil_2001)
print("Biner hasil =", bin(hasil_2001))
print("Biner hasil (8 bit) =", format(hasil_2001, "08b"))

# Bitwise OR
hasil_2001 = angka1_2001 | angka2_2001
print("\nBitwise OR (|)")
print(angka1_2001, "|", angka2_2001, "=", hasil_2001)
print("Biner hasil =", bin(hasil_2001))
print("Biner hasil (8 bit) =", format(hasil_2001, "08b"))

# Bitwise XOR
hasil_2001 = angka1_2001 ^ angka2_2001
print("\nBitwise XOR (^)")
print(angka1_2001, "^", angka2_2001, "=", hasil_2001)
print("Biner hasil =", bin(hasil_2001))
print("Biner hasil (8 bit) =", format(hasil_2001, "08b"))

# Bitwise NOT
hasil_2001 = ~angka1_2001
print("\nBitwise NOT (~)")
print("~", angka1_2001, "=", hasil_2001)
print("Biner hasil =", bin(hasil_2001))
print("Biner hasil (8 bit) =", format(hasil_2001, "08b"))

# Bitwise geser kiri
jumlah_geser_2001 = int(input("\nMasukkan jumlah pergeseran bit: "))

hasil_2001 = angka1_2001 << jumlah_geser_2001
print("\nBitwise geser kiri (<<)")
print(angka1_2001, "<<", jumlah_geser_2001, "=", hasil_2001)
print("Biner hasil =", bin(hasil_2001))
print("Biner hasil (8 bit) =", format(hasil_2001, "08b"))

# Bitwise geser kanan
hasil_2001 = angka1_2001 >> jumlah_geser_2001
print("\nBitwise geser kanan (>>)")
print(angka1_2001, ">>", jumlah_geser_2001, "=", hasil_2001)
print("Biner hasil =", bin(hasil_2001))
print("Biner hasil (8 bit) =", format(hasil_2001, "08b"))
```

**Penjelasan :**

**1. Membuat Komentar dan Judul Header:** Sintaks diawali tanda `#` (seperti `#` Program ini menggunakan fungsi `input())` digunakan sebagai komentar penjelas alur program. Sintaks `print()` header digunakan untuk menampilkan judul modul "3. OPERATOR BITWISE" agar tampilan antarmuka output terstruktur.

**2. ​Menerapkan Input Data dan Konversi Representasi Biner:** Sintaks `angka1_2001 = int(input("..."))` dan `angka2_2001 = int(input("..."))` mengambil masukan angka desimal dari pengguna dan mengonversinya menjadi integer. Fungsi `bin()` digunakan untuk menampilkan representasi nilai biner dasar (diawali penanda `0b`), sedangkan fungsi `format(hasil_2001, "08b")` memformat tampilan nilai biner agar selalu memiliki panjang 8-bit.

**3. ​Menerapkan Operator Bitwise Logika (AND, OR, XOR, NOT):**

> * ​Sintaks `&` (AND) membandingkan tiap bit dan menghasilkan nilai 1 jika kedua bit bernilai 1.

> * ​Sintaks `|` (OR) menghasilkan nilai 1 jika salah satu bit bernilai 1.

> * ​Sintaks `^` (XOR) menghasilkan nilai 1 hanya jika kedua bit memiliki nilai yang berbeda.

> * ​Sintaks `~` (NOT) melakukan pembalikan bit (bitwise complement), di mana nilai `~n` dihitung berdasarkan rumus `-(n + 1)`.

**4. ​Menerapkan Operator Bitwise Shift (Geser Kiri dan Geser Kanan):** Sintaks `jumlah_geser_2001` mengambil masukan jumlah posisi pergeseran bit. Operator `<<` (Left Shift) menggeser bit `angka1_2001` ke kiri sebanyak n posisi (setara dengan mengalikan angka dengan 2^n). Sedangkan operator `>>` (Right Shift) menggeser bit ke kanan sebanyak n posisi (setara dengan pembagian bulat angka dengan 2^n).

**5. ​Mengeluarkan Output Program:** Perintah `print()` menampilkan hasil kalkulasi angka desimal, format biner standar `(bin())`, dan format biner 8-bit `(format())` untuk setiap pengujian operator bitwise ke layar output terminal.

**​4. Perbedaan Antara Operator `is` dan `==` pada Python**

> * **​Operator `==` (Perbandingan Nilai):* Membandingkan apakah dua variabel memiliki isi atau nilai data yang sama, tanpa memedulikan alamat memorinya.

_​Contoh:_ `total_2001 == 200000` memeriksa apakah nilai variabel `total_2001` sama dengan 200.000.

> * **​Operator `is` (Identitas Objek/Memori):** Membandingkan apakah dua variabel merujuk pada objek memori (memory address) yang sama persis di dalam sistem.

_​Contoh:_ `cek_tipe_2001 = type(bayar_2001) is float` memeriksa apakah objek tipe data dari `bayar_2001` merupakan tipe float secara identitas kelas di memori.

**​5. Penggunaan Operator Bitwise untuk Menentukan Hak Akses Pelanggan**

​Operator bitwise digunakan melalui teknik bitwise flag (penanda bit) 4-bit biner:

> * ​Setiap syarat transaksi diwakili oleh bit tertentu: Member (`0b0001`), Total Belanja (`0b0010`), Jumlah Barang (`0b0100`), dan Kode Promo (`0b1000`).

> * ​Operator OR (`|`): Digunakan untuk menggabungkan seluruh kondisi hak akses yang terpenuhi menjadi satu nilai biner gabungan (`bit_status_2001`).

> * ​Operator AND (`&`): Digunakan untuk melakukan masking atau pengecekan ulang bit mana saja yang bernilai aktif (1) pada status transaksi.

> * ​Operator XOR (`^`) & Shift (`<<`): XOR membandingkan perbedaan status biner transaksi dengan bit referensi penuh (`0b1111`), sedangkan Left Shift menggeser posisi biner untuk kalkulasi hak akses berbasis bit manipulation.

**​6. Alur Program dari Masukan (Input) hingga Output Pembayaran dan Promo**

`[ Input Data ] ---> [ Normalisasi Teks & Cast ] ---> [ Evaluasi Kondisi & Diskon ] ---> [ Bitwise & Final Output ]`

> * **​Penerimaan & Normalisasi Input:** Program meminta masukan nama, status pelanggan, total belanja, jumlah barang, dan kode promo. Teks status diubah ke huruf kecil (`.lower()`) dan kode promo diubah ke kapital (`.upper()`).

> * **​Validasi Kondisi Diskon:** Program memeriksa kelayakan diskon via kombinasi operator perbandingan dan logika.

> * **​Kalkulasi Pembayaran:** Struktur percabangan `if-elif-else` menentukan besarnya potongan harga. Program mengalkulasi total bayar (`total - diskon`), rata-rata harga per barang, serta sisa barang (`% 3`).

> * **​Pemrosesan Status Bitwise & Output:** Program membentuk kode status biner 4-bit menggunakan operator bitwise, lalu seluruh rincian transaksi, status promo, dan kode biner dicetak ke konsol.

**​7. Identifikasi Kondisi Program yang Menghasilkan True dan False**

> * ​Kondisi yang menghasilkan True:

`​kode_2001 in daftar_kode_2001`

(Bernilai True jika masukan pengguna adalah `"HEMAT10"`, `"HEMAT20"`, atau `"GRATISONGKIR"`).

> * ​Kondisi yang menghasilkan False:

`​kode_2001 not in daftar_kode_2001`

(Bernilai False jika masukan pengguna terdaftar resmi di dalam list promo).

<h1><b>KESIMPULAN</b></h1>

</div>

Berdasarkan praktikum mengenai operator pada bahasa pemrograman Python, dapat ditarik beberapa kesimpulan sebagai berikut:

* **​Fungsi Operator Aritmetika dan Penugasan (Assignment):**

Operator aritmetika (`+`, `-`, `*`, `/`, `//`, `%`, `**`) berfungsi untuk menangani pengolahan matematis seperti menghitung nilai transaksi, diskon, dan rata-rata. Sementara itu, operator penugasan gabungan (seperti `+=`, `-=`, `*=`) menyederhanakan pembaruan nilai variabel secara efisien.

* **​Evaluasi Relasional dan Logika:**

Operator perbandingan (`==`, `!=`, `>`, `<`, `>=`, `<=`) menghasilkan nilai logika bertipe boolean (True atau False). Penggabungan beberapa kondisi logika menggunakan operator and, or, dan not memungkinkan pembuatan kriteria kelayakan sistem (seperti hak diskon pelanggan) yang kompleks dan presisi.

* **​Pengujian Keanggotaan dan Identitas Objek:**

Operator keanggotaan (`in` dan `not in`) terbukti efektif untuk memvalidasi ketersediaan data pada koleksi list (seperti pengecekan kode promo). Operator identitas (`is` dan `is not`) digunakan untuk membandingkan alamat lokasi memori dari dua objek, yang berbeda prinsipnya dengan perbandingan nilai (`==`).

* **​Manipulasi Data Tingkat Bit (Bitwise):**

Operator bitwise (`&`, `|`, `^`, `~`, `<<`, `>>`) mengolah data pada tingkat biner (0 dan 1). Teknik bitwise flag memungkinkan beberapa syarat status transaksi atau hak akses dikodekan secara ringkas ke dalam bentuk status biner 4-bit.
