<h2><b>2.1 Landasan Teori</b></h2>

Landasan teori ini menguraikan beberapa konsep dan perangkat utama yang digunakan dalam pelaksanaan praktikum Algoritma dan Pemrograman, antara lain:

**1. ​Bahasa Pemrograman Python**

Python adalah bahasa pemrograman tingkat tinggi yang bersifat interpreter, dinamis, dan multiguna (general-purpose). Python dibuat oleh Guido van Rossum dan dirilis pada tahun 1991. Python dirancang dengan filosofi yang mengutamakan tingkat keterbacaan kode (code readability). Sintaksnya yang bersih dan menyerupai bahasa Inggris membuat Python sangat ramah bagi pemula, namun tetap memiliki performa yang kuat untuk menangani proyek skala besar, pengembangan web, kecerdasan buatan (AI), hingga _data science_.

**2. ​Visual Studio Code (VS Code)**

Visual Studio Code adalah sebuah _source-code editor_ ringan namun powerful yang dikembangkan oleh Microsoft untuk sistem operasi Windows, Linux, dan macOS. VS Code dilengkapi dengan fitur bawaan seperti dukungan debugging, kontrol Git terintegrasi, penyorotan sintaks (syntax highlighting), penyelesaian kode cerdas (IntelliSense), serta kemampuan ekstensi yang sangat luas. Dalam praktikum pemrograman, VS Code sering digunakan bersama ekstensi bahasa Python untuk memudahkan proses penulisan, pengujian, dan eksekusi kode program.

**3. ​Git dan GitHub**

Git adalah sistem pengontrol versi terdistribusi (distributed version control system) yang berfungsi untuk melacak setiap perubahan pada kode sumber selama proses pengembangan perangkat lunak berlangsung.

GitHub adalah platform berbasis web yang menggunakan Git untuk manajemen repositori jarak jauh (remote repository). GitHub memungkinkan pengembang untuk menyimpan kode secara daring, melakukan pencatatan perubahan melalui perintah _commit_, mencadangkan proyek, serta berkolaborasi dengan tim pengembang lain dari berbagai belahan dunia secara _real-time_.

<h2><b>2.2 Koneksi ke GitHub</b></h2>

Untuk menghubungkan repositori lokal di komputer ke GitHub serta melakukan _commit_ dan _push_ secara aman menggunakan token, lakukan tahapan berikut:

**1. Mengakses Direktori Berkas / Proyek di Terminal**

Buka terminal di VS Code, lalu arahkan direktori aktif ke folder tempat penyimpanan file praktikum Anda menggunakan perintah `cd` (_change directory_):

```python
cd "path/ke/folder/proyek"
```

Contoh pada sistem Linux di VS Code:

```python
cd "C:/Nama_Folder_Praktikum"
```

**Penjelasan:** Perintah ini memastikan terminal berada tepat di dalam folder proyek yang ingin diunggah ke GitHub.

**2. Inisialisasi Git pada Direktori Proyek**

Setelah berada di dalam folder proyek, jalankan perintah berikut untuk menginisialisasi repositori Git lokal:

```python
git init
```

**Penjelasan:** Perintah ini membuat folder tersembunyi `.git` yang berfungsi agar sistem Git mulai memantau dan melacak perubahan pada berkas-berkas di dalam direktori tersebut.

**3. Menghubungkan Repositori Lokal ke GitHub (Remote Repository)**

Daftarkan tautan repositori jarak jauh (remote) yang telah dibuat sebelumnya di GitHub:

```python
git remote add origin https://github.com/username_github/nama_repo.git
```

**Penjelasan:** Perintah ini berfungsi untuk menghubungkan folder lokal perangkat dengan repositori tujuan di server GitHub menggunakan protokol HTTPS.

**4. Menyiapkan Berkas dan Melakukan Commit**

Masukkan seluruh berkas ke dalam area penyiapan (staging area), lalu buat catatan perubahan (commit):

```python
git add .
git commit -m "pesan_commit"
```

Penjelasan: `git add .` memilih semua berkas yang akan disimpan, sedangkan `git commit -m` mencatat versi perubahan tersebut secara lokal di dalam komputer disertai dengan pesan deskriptif.

**5. Menyesuaikan Nama Branch Utama**

Pastikan nama branch lokal sudah sesuai dengan standar branch di GitHub (menggunakan main atau master):

```python
git branch -M main
```

**6. Mengirim Kode ke GitHub (Push) Menggunakan Personal Access Token (PAT)**

Kirimkan kode yang telah di-commit ke repositori GitHub dengan perintah:

```python
git push -u origin main
```

**Proses Autentikasi (Keamanan Komputer Lab):**

> * Saat terminal meminta Username, masukkan username akun GitHub.

> * Saat terminal meminta Password, jangan masukkan kata sandi akun GitHub, melainkan masukkan Personal Access Token (PAT) yang telah dibuat sebelumnya melalui menu `Settings > Developer Settings > Personal access tokens` di GitHub.

<h3><b>2.3.1 Program "A_2611532001"</b></h3>

```python
# Program pertama
"""
Komentar
dua baris atau lebih
pertemuan 1 praktikum alpro
"""
print("Hello World")
print()
print("Program ini menghasilkan")
print("Empat baris output")
print(r"/ \\ // \\\ /// \\\\")
print("program ini" + " dibuat oleh a/n "
+ "Rizki Ramadhani Al Michtari dan 2611532001.")
```

**Penjelasan :**
1. **Membuat Komentar 1 Baris:** Pada sintaks
`# Program Pertama`
terdapat simbol `#` yang berfungsi untuk memberikan komentar untuk 1 baris kode. Jika ingin menambahkan komentar lagi dengan simbol yang sama, maka di baris berikutnya harus diawali juga dengan simbol `#`.
2. **Membuat Komentar Lebih dari 1 Baris:** Pada sintaks
`"""`
`Komentar`
`dua baris atau lebih`
`pertemuan 1 praktikum alpro`
`"""`
terdapat simbol 3 buah petik ganda (`"""`) yang berfungsi untuk memberikan komentar di kode program yang lebih dari 1 baris. Cara ini lebih efisien daripada menggunakan simbol `#` untuk membuat komentar dengan lebih dari 1 baris.

Note : Program Python tidak akan mengeluarkan output apapun dari komentar yang ada di kodenya. Komentar merupakan suatu sintak kode yang digunakan untuk memberikan penjelasan atau catatan yang tidak akan dieksekusi oleh program.

3. **Mengeluarkan _Output_ pada Program Python:** Pada sintaks
`print("Hello World")`
terdapat fungsi `print` yang berfungsi untuk mengeluarkan _output_ `Hello World`. Dari sintaks `print("Hello World")`. Selain mengeluarkan output _string_, fungsi `print` juga bisa digunakan untuk mengeluarkan _output_ nilai dari suatu _variabel_ di dalam program. Contoh, jika terdapat sintaks `x=5  print(x)`, maka program akan mengeluarkan nilai dari `x` yaitu `5`.
4. **Menggabungkan String:** Pada sintaks
`print("program ini" + " dibuat oleh a/n "`
`+ "Rizki Ramadhani Al Michtari dan 2611532001.")`
terdapat simbol `+` yang berfungsi untuk mengeluarkan _output_ yang menggabungkan 2 atau lebih _string_. _String_ `program ini` dan _string_ ` dibuat oleh a/n` digabung menjadi 1 kalimat pada _output_ program kode Python, sehingga menjadi `program ini dibuat oleh a/n`, begitu pula dengan _string_ `Rizki Ramadhani Al Michtari dan 2611532001.`
