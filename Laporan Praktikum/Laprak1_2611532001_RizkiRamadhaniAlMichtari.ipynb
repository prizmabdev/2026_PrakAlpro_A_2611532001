### 2.2.1 Program "A_2611532001"

```python
# Program pertama
"""
Komentar
dua baris atau lebih
pertemuan 1 praktikum alpro
"""
print("Hello World")
print()
print("Program ini menghasilkan")
print("Empat baris output")
print("/ \\ // \\\ /// \\\\")
print("program ini" + " dibuat oleh a/n "
+ "Rizki Ramadhani Al Michtari dan 2611532001.")
```

**Penjelasan :**
1. **Membuat Komentar 1 Baris:** Pada sintaks
`# Program Pertama`
terdapat simbol `#` yang berfungsi untuk memberikan komentar untuk 1 baris kode. Jika ingin menambahkan komentar lagi dengan simbol yang sama, maka di baris berikutnya harus diawali juga dengan simbol `#`.
2. **Membuat Komentar Lebih dari 1 Baris:** Pada sintaks
`"""`
`Komentar`
`dua baris atau lebih`
`pertemuan 1 praktikum alpro`
`"""`
terdapat simbol 3 buah petik ganda (`"""`) yang berfungsi untuk memberikan komentar di kode program yang lebih dari 1 baris. Cara ini lebih efisien daripada menggunakan simbol `#` untuk membuat komentar dengan lebih dari 1 baris.

Note : Program Python tidak akan mengeluarkan output apapun dari komentar yang ada di kodenya. Komentar merupakan suatu sintak kode yang digunakan untuk memberikan penjelasan atau catatan yang tidak akan dieksekusi oleh program.

3. **Mengeluarkan _Output_ pada Program Python:** Pada sintaks
`print("Hello World")`
terdapat fungsi `print` yang berfungsi untuk mengeluarkan _output_ `Hello World`. Dari sintaks `print("Hello World")`. Selain mengeluarkan output _string_, fungsi `print` juga bisa digunakan untuk mengeluarkan _output_ nilai dari suatu _variabel_ di dalam program. Contoh, jika terdapat sintaks `x=5  print(x)`, maka program akan mengeluarkan nilai dari `x` yaitu `5`.
4. **Menggabungkan String:** Pada sintaks
`print("program ini" + " dibuat oleh a/n "`
`+ "Rizki Ramadhani Al Michtari dan 2611532001.")`
terdapat simbol `+` yang berfungsi untuk mengeluarkan _output_ yang menggabungkan 2 atau lebih _string_. _String_ `program ini` dan _string_ ` dibuat oleh a/n` digabung menjadi 1 kalimat pada _output_ program kode Python, sehingga menjadi `program ini dibuat oleh a/n`, begitu pula dengan _string_ `Rizki Ramadhani Al Michtari dan 2611532001.`
