<h2><b>2.1 Landasan Teori</b></h2>

Di dalam membangun sebuah program komputer, struktur kontrol alur sangat diperlukan agar eksekusi perintah tidak hanya berjalan lurus dari atas ke bawah secara kaku. Salah satu mekanisme kontrol alur yang paling sering digunakan adalah perulangan (_looping_), yang memungkinkan kita untuk menjalankan blok kode yang sama secara berulang-ulang tanpa harus menulisnya berkali-kali. Di dalam bahasa pemrograman Python, salah satu instrumen perulangan utama yang diandalkan adalah perulangan for (`for loop`).

### **2.1.1 Konsep Dasar Perulangan For (For Loop)**

Secara mendasar, perulangan for di dalam Python dirancang khusus untuk menelusuri atau mengiterasi setiap elemen yang ada di dalam sebuah objek kumpulan data atau objek yang dapat diulang (_iterable_), seperti deret angka, karakter di dalam string, maupun elemen di dalam struktur data lainnya. Ketika program menemui pernyataan _for_, variabel penampung (_iterator_) akan secara otomatis mengambil nilai dari setiap elemen satu per satu secara berurutan, lalu mengeksekusi blok kode di dalam indentasi hingga seluruh elemen habis diproses.

### **2.1.2 Pemanfaatan Fungsi `range()` sebagai Pembangkit Deret Angka**

Dalam banyak kasus pemrograman, perulangan _for_ sering kali dipadukan dengan fungsi bawaan Python bernama `range()`. Fungsi ini berfungsi untuk menghasilkan rentang atau deret bilangan bulat secara otomatis tanpa kita harus mengetikannya satu persatu. Berdasarkan panduan resmi Python, fungsi `range()` ini cukup fleksibel karena dapat menerima satu, dua, atau tiga argumen sekaligus untuk mengatur jalannya iterasi:

> * **Satu Parameter (`range(stop)`):** Digunakan jika kita ingin memulai hitungan secara default dari angka 0 hingga tepat satu angka sebelum nilai stop yang ditentukan.

> * **Dua Parameter (`range(start, stop)`):** Digunakan ketika kita ingin menentukan titik awal hitungan secara spesifik melalui parameter start, berjalan naik hingga sebelum angka stop.

> * **Tiga Parameter (range(`start, stop, step)`):** Digunakan untuk mengatur jarak atau kelipatan lompatan antar angka iterasi melalui parameter step, yang memungkinkan hitungan maju maupun mundur secara terstruktur.

### **2.1.3 Perulangan Bersarang (_Nested For Loop_)**

Dalam penyelesaian masalah komputasi yang lebih kompleks, seperti pengolahan matriks dua dimensi, pembuatan tabel aritmatika, ataupun pencetakan pola geometri berbasis karakter, satu buah perulangan saja sering kali tidak cukup. Di sinilah konsep perulangan bersarang (_nested for loop_) diterapkan. _Nested for loop_ adalah kondisi di mana terdapat struktur perulangan _for_ yang berada di dalam blok perulangan for lainnya. Cara kerjanya, untuk setiap satu kali putaran penuh pada perulangan luar (_outer loop_), perulangan bagian dalam (_inner loop_) akan menyelesaikan seluruh siklus iterasinya dari awal hingga akhir.

### **2.1.4 Pengendalian Tampilan Konsol dan Akumulasi Data**

Dalam pengoperasian perulangan di dalam konsol teks, Python secara otomatis memindahkan kursor ke baris baru setiap kali perintah `print()` selesai dijalankan. Untuk kebutuhan pencetakan pola matriks atau deret mendatar (_horizontal_), kita dapat memanfaatkan argumen tambahan berupa `end=" "` pada fungsi cetak guna mencegah perpindahan baris tersebut. Selain itu, perulangan for juga sangat esensial untuk proses komputasi yang melibatkan akumulasi nilai secara progresif, seperti penjumlahan deret angka atau perhitungan total transaksi dengan cara memanfaatkan variabel penampung (_accumulator_) yang nilainya terus diperbarui pada setiap putaran iterasi.

<h2><b>2.2 Kode Program</b></h2>

<h3><b>2.2.1 Program "nested_for1_2611532001.py"</b></h3>

```python
# Buat program untuk perulangan for dalam python
# Nama variabel ditambah 4 digit nim terakhir contoh: ulang_1234
# Program ini menggunakan fungsi input()

batas_2001 = int(input("Masukkan nilai batas: "))
for line_2001 in range(1, batas_2001 + 1):
    for j_2001 in range(1, (-1 * line_2001 + batas_2001) + 1):
        print(".", end=" ")
    print(line_2001)
```

**Penjelasan :**

**1. Membuat Komentar dan Menerima Masukan Batas Angka:** Tanda `#` digunakan sebagai komentar penjelas alur program. Sintaks `input()` menangkap masukan nilai batas dari pengguna, yang kemudian dikonversi secara eksplisit (_type casting_) menggunakan fungsi `int()` menjadi bilangan bulat (_integer_) pada `variabel batas_2001`. Nilai ini nantinya akan menentukan ukuran matriks atau tinggi pola yang dicetak ke layar.

**2. Menerapkan Perulangan Luar:** Pernyataan `for line_2001 in range(1, batas_2001 + 1):` mengontrol jumlah baris yang akan dicetak secara vertikal. Fungsi `range(1, batas_2001 + 1)` menghasilkan deret angka berurutan mulai dari 1 hingga nilai `batas_2001`, di mana setiap iterasi mewakili nomor baris yang sedang aktif pada variabel `line_2001`.

**3. Menerapkan Perulangan Dalam untuk Karakter Titik:** Di dalam setiap baris, terdapat perulangan kedua yaitu `for j_2001 in range(1, (-1 * line_2001 + batas_2001) + 1):`. Perulangan ini bertugas mencetak karakter titik (`.`) secara berulang ke samping. Rumus batas atas rentang `(-1 * line_2001 + batas_2001)` dirancang secara dinamis agar jumlah titik yang dicetak pada setiap baris berkurang secara bertahap seiring bertambahnya nomor baris.

**4. Argumen `end=" "` pada Fungsi Cetak Titik:** Perintah `print(".", end=" ")` digunakan untuk mencetak karakter titik diikuti dengan spasi, sekaligus mencegah perpindahan baris otomatis. Hal ini membuat seluruh karakter titik pada baris tersebut tercetak secara mendatar (horizontal) di dalam satu baris konsol yang sama.

**5. Mencetak Nomor Baris di Ujung Baris:** Perintah `print(line_2001)` diletakkan di dalam perulangan luar namun di luar perulangan dalam. Perintah ini berfungsi untuk mencetak nilai variabel `line_2001` sekaligus bertindak sebagai pemindah baris otomatis, sehingga pola baris berikutnya akan tercetak di bawahnya secara rapi membentuk pola segitiga terbalik.

<h3><b>2.2.2 Program "nested_for2_2611532001.py"</b></h3>

```python
# Buat program untuk perulangan for dalam python
# Nama variabel ditambah 4 digit nim terakhir contoh: ulang_1234
# Program ini menggunakan fungsi input()

batas_2001 = int(input("Masukkan nilai batas: "))
for i_2001 in range(1, batas_2001 + 1):
    for j_2001 in range(1, batas_2001 + 1):
        print("*", end=" ")
    print() # pindah ke baris berikutnya
```

**Penjelasan :**

**1. Penerimaan Masukan Nilai Batas:** Tanda `#` berfungsi sebagai komentar penjelas. Sintaks `input()` mengambil masukan angka dari pengguna secara interaktif, yang kemudian dikonversi secara eksplisit (_type casting_) menggunakan fungsi `int()` menjadi bilangan bulat pada variabel `batas_2001`. Nilai ini menentukan dimensi ukuran panjang dan lebar dari pola matriks yang akan dibentuk.

**2. Mengontrol Perulangan Baris Luar:** Pernyataan `for i_2001 in range(1, batas_2001 + 1):` berfungsi sebagai perulangan utama untuk mengendalikan pencetakan baris secara vertikal dari atas ke bawah. Fungsi `range(1, batas_2001 + 1)` menghasilkan deret angka mulai dari 1 hingga `nilai batas_2001`.

**3. Mengontrol Perulangan Kolom Dalam:** Di dalam perulangan baris, terdapat perulangan kedua yaitu `for j_2001 in range(1, batas_2001 + 1):`. Perulangan ini berjalan secara penuh dari 1 hingga `batas_2001` pada setiap satu kali iterasi baris, yang bertugas mencetak karakter bintang secara berulang ke arah samping (_horizontal_).

**4. Argumen `end=" "` pada Pencetakan Karakter:** Perintah `print("*", end=" ")` digunakan untuk mencetak simbol bintang diikuti oleh spasi, tanpa memindahkan kursor ke baris baru. Hal ini memastikan seluruh bintang pada baris yang sama tersusun secara mendatar di konsol.

**5. Pemindahan Baris Otomatis:** Perintah `print()` kosong yang diletakkan di dalam perulangan luar (di luar perulangan dalam) berfungsi untuk menghasilkan karakter baris baru. Perintah ini dieksekusi setiap kali satu baris matriks selesai dicetak sepenuhnya, sehingga iterasi baris berikutnya dapat berlanjut ke bawah dan membentuk bangun datar persegi yang rapi.

<h3><b>2.2.3 Program "nested_for3_2611532001.py"</b></h3>

```python
# Buat program untuk perulangan for dalam python
# Nama variabel ditambah 4 digit nim terakhir contoh: ulang_1234
# Program ini menggunakan fungsi input()

batas_2001 = int(input("Masukkan nilai batas: "))
for i_2001 in range(batas_2001 + 1):
    for j_2001 in range(batas_2001 + 1):
        print(i_2001 + j_2001, end=" ")
    print() # pindah ke baris berikutnya
```

**Penjelasan :**

**1. Penerimaan Masukan Nilai Batas:** Tanda `#` digunakan sebagai komentar penjelas. Nilai ini menentukan batas maksimum rentang perulangan matriks.

**2. Mengontrol Perulangan Baris Luar:** Pernyataan `for i_2001 in range(batas_2001 + 1):` berfungsi sebagai perulangan utama untuk mengendalikan baris secara vertikal. Karena nilai awal pada fungsi `range()` tidak ditentukan secara eksplisit, fungsi ini otomatis dimulai dari angka 0 hingga `batas_2001` (inklusif karena ditambah 1).

**3. Mengontrol Perulangan Kolom Dalam:** Pernyataan `for j_2001 in range(batas_2001 + 1):` bertindak sebagai perulangan bagian dalam yang berjalan penuh dari angka 0 hingga `batas_2001` untuk setiap satu kali putaran baris, guna mengisi kolom secara mendatar (horizontal).

**4. Operasi Penjumlahan Variabel Loop dan Argumen `end=" "`:** Di dalam perulangan, perintah `print(i_2001 + j_2001, end=" ")` mengevaluasi hasil penjumlahan nilai variabel baris (`i_2001`) dan variabel kolom (`j_2001`) pada setiap iterasi. Hasil penjumlahan tersebut dicetak ke layar diikuti dengan spasi tanpa memindahkan kursor ke baris baru, sehingga membentuk format matriks angka yang rapi.

**5. Pemindahan Baris Otomatis:** Perintah `print()` kosong yang diletakkan di luar perulangan dalam berfungsi untuk memasukkan karakter baris baru. Perintah ini dieksekusi setelah satu baris penuh selesai dicetak, sehingga iterasi baris berikutnya turun ke bawah membentuk tabel matriks aritmatika yang utuh.

<h3><b>2.2.4 Program "nested_for4_2611532001.py"</b></h3>

```python
# Buat program untuk perulangan for dalam python
# Nama variabel ditambah 4 digit nim terakhir contoh: ulang_1234
# Program ini menggunakan fungsi input()

tinggi_2001 = int(input("Masukkan tinggi (bilangan genap, misal 10): "))

if tinggi_2001 % 2 != 0:
    print("Tinggi harus bilangan genap!")
else:
    a_2001 = tinggi_2001
    c_2001 = a_2001
    lebar_2001 = (2 * tinggi_2001) - 2

    for i_2001 in range(1, tinggi_2001 + 1):
        b_2001 = c_2001 + 1

        for j_2001 in range(1, lebar_2001 + 1):
            # Baris atas dan bawah
            if i_2001 == 1 or i_2001 == tinggi_2001:
                if j_2001 == 1 or j_2001 == lebar_2001:
                    print("#", end="")
                else:
                    print("=", end="")
            # Baris isi
            else:
                if j_2001 == 1 or j_2001 == lebar_2001:
                    print("|", end="")
                else:
                    if j_2001 == c_2001:
                        print("<", end="")
                    elif j_2001 == b_2001:
                        print(">", end="")
                    elif j_2001 == (lebar_2001 - c_2001):
                        print("<", end="")
                    elif j_2001 == (lebar_2001 - c_2001 + 1):
                        print(">", end="")
                    elif j_2001 > b_2001 and j_2001 < (lebar_2001 - c_2001):
                        print(".", end="")
                    else:
                        print(" ", end="")
        print()

        # Logika asli Java
        a_2001 -= 2

        if a_2001 <= 0:
            c_2001 = (-a_2001) + 2
        else:
            c_2001 = a_2001
```

**Penjelasan :**

**1. Inisialisasi Variabel Tinggi dan Validasi Awal Bilangan Genap:** Tanda `#` digunakan sebagai komentar penjelas. Sintaks `input()` mengambil masukan angka dari pengguna yang dikonversi ke tipe bilangan bulat pada variabel `tinggi_2001`. Program kemudian menggunakan struktur percabangan `if (tinggi_2001 % 2 != 0):` untuk memeriksa apakah nilai tinggi yang dimasukkan adalah bilangan ganjil. Jika terbukti ganjil, program mencetak pesan peringatan dan menghentikan alur eksekusi, karena pola geometri ini dirancang khusus untuk parameter angka genap.

**2. Inisialisasi Parameter Lebar dan Variabel Pembantu:** Jika masukan bernilai genap, program menginisialisasi beberapa variabel pengendali seperti `a_2001`, `c_2001`, serta menghitung total lebar matriks berdasarkan rumus `lebar_2001 = (2 * tinggi_2001) - 2`. Variabel-variabel ini nantinya akan bertindak sebagai acuan posisi koordinat karakter di dalam bidang cetak dua dimensi.

**3. Pengendalian Perulangan Baris Vertikal:** Pernyataan `for i_2001 in range(1, tinggi_2001 + 1):` berfungsi sebagai perulangan utama untuk mengendalikan proses pencetakan baris dari atas ke bawah. Di setiap awal iterasi baris, variabel `b_2001` diperbarui berdasarkan `nilai c_2001 + 1` untuk menentukan posisi batas koordinat karakter berikutnya.

**4. Pengendalian Perulangan Kolom Horizontal dan Validasi Bingkai Atas-Bawah:** Di dalam perulangan baris, terdapat perulangan kedua yaitu `for j_2001 in range(1, lebar_2001 + 1):` yang menyusuri setiap kolom secara mendatar. Melalui percabangan di dalamnya, jika program mendeteksi posisi baris pertama atau terakhir (`i_2001 == 1 or i_2001 == tinggi_2001`), maka sudut bingkai akan dicetak menggunakan simbol pagar (`#`), sementara sisi horizontal bagian atas dan bawah dicetak menggunakan simbol tanda sama dengan (`=`).

**5. Pencetakan Bingkai Samping dan Pola Karakter Interior:** Untuk baris-baris di bagian tengah, sisi paling kiri dan kanan bingkai dicetak menggunakan karakter garis tegak (`|`). Di bagian interiornya, program menerapkan percabangan bertingkat yang sangat presisi untuk menempatkan simbol panah kiri (`<`), panah kanan (`>`), titik di bagian tengah (`.`), serta spasi kosong secara dinamis berdasarkan perbandingan posisi variabel koordinat kolom `j_2001` terhadap batas variabel `c_2001` dan `b_2001`.

**6. Pemindahan Baris dan Pembaruan Logika Iterasi (Decrement/Increment):** Setiap kali satu baris selesai diproses oleh perulangan kolom, perintah `print(`) kosong dieksekusi untuk memindahkan kursor ke baris baru. Di akhir iterasi baris, nilai variabel pembantu `a_2001` dikurangi dua (`a_2001 -= 2`), dan menggunakan struktur kondisional sederhana, nilai `c_2001` disesuaikan kembali secara matematis untuk mempersiapkan pergeseran pola simetris pada baris berikutnya hingga mencapai batas tinggi maksimum.

<h3><b>2.2.5 Program "perulangan_for1_2611532001.py"</b></h3>

```python
# Buat program untuk perulangan for dalam python
# Nama variabel ditambah 4 digit nim terakhir contoh: ulang_1234
# Program ini menggunakan fungsi input()

ulang_2001 = int(input("Masukkan Jumlah Perulangan: "))

for i_2001 in range (ulang_2001):
    print(f"Perulangan ke-{i_2001}")
```

**Penjelasan :**

**1. Menerima Masukan Jumlah Perulangan:** Tanda `#` digunakan sebagai komentar penjelas. Sintaks `input()` mengambil masukan angka dari pengguna secara interaktif, yang kemudian dikonversi secara eksplisit (_type casting_) menggunakan fungsi `int()` menjadi bilangan bulat pada variabel `ulang_2001`. Nilai ini menentukan berapa kali total perulangan akan dieksekusi oleh program.

**2. Penerapan Struktur Perulangan for dan Fungsi `range()`:** Pernyataan `for i_2001 in range(ulang_2001):` bertindak sebagai pengendali utama iterasi. Penggunaan fungsi `range(ulang_2001)` secara default akan menghasilkan deretan angka berurutan yang dimulai dari angka 0 hingga bilangan sebelum nilai `ulang_2001` (sebanyak jumlah masukan yang diberikan).

**3. Penggunaan Variabel Iterator:** Pada setiap putaran iterasi, `variabel i_2001` otomatis menampung nilai angka urutan iterasi yang sedang berjalan (bernilai dari 0 hingga `ulang_2001 - 1`).

**4. Pencetakan Output Berformat:** Di dalam blok perulangan, perintah `print(f"Perulangan ke-{i_2001}")` menggunakan fitur f-string untuk menyisipkan nilai variabel `i_2001` secara langsung ke dalam teks string, sehingga konsol dapat menampilkan urutan status iterasi secara dinamis pada setiap barisnya.

<h3><b>2.2.6 Program "perulangan_for2_2611532001.py"</b></h3>

```python
# Buat program untuk perulangan for dalam python
# Nama variabel ditambah 4 digit nim terakhir contoh: ulang_1234
# Program ini menggunakan fungsi input()

ulang_2001 = int(input("Masukkan jumlah perulangan: "))

print("Perulangan ke-0 sampai ke-", ulang_2001 - 1)
for i_2001 in range(ulang_2001):
    print(i_2001, end=" ")
print()

print("Perulangan ke-1 sampai ke-", ulang_2001)
for i in range(1, ulang_2001 + 1):
    print(i, end=" ")
```

**Penjelasan:**

**1. Menerima Masukan Jumlah Perulangan:** Tanda `#` digunakan sebagai komentar penjelas. Sintaks `input()` menangkap masukan angka dari pengguna, yang kemudian dikonversi secara eksplisit menggunakan fungsi `int()` menjadi bilangan bulat pada variabel `ulang_2001`. Nilai ini menjadi batas akhir atau jumlah iterasi yang akan dieksekusi.

**2. Mencetak Deret Angka Mulai dari Nol:** Perintah `print("Perulangan ke-0 sampai ke-", ulang_2001 - 1)` menampilkan informasi teks pendahuluan. Di bawahnya, perulangan `for i_2001 in range(ulang_2001):` menggunakan fungsi `range()` dengan satu parameter, yang secara otomatis menghasilkan urutan angka mulai dari 0 hingga angka sebelum `ulang_2001`.

**3. Pencetakan Mendatar Menggunakan Argumen `end=" "`:** Di dalam blok perulangan pertama, perintah `print(i_2001, end=" ")` mencetak setiap nilai iterator `i_2001` secara mendatar dengan pemisah spasi tanpa memindahkan kursor ke baris baru. Perintah `print()` kosong setelah perulangan berfungsi untuk memberikan baris baru.

**4. Mencetak Deret Angka dengan Parameter Awal Khusus (`range(start, stop)`):** Perintah `print("Perulangan ke-1 sampai ke-", ulang_2001)` mencetak informasi teks untuk bagian kedua. Selanjutnya, perulangan `for i in range(1, ulang_2001 + 1):` menggunakan dua parameter pada fungsi `range()`, yaitu angka awal 1 dan batas akhir `ulang_2001 + 1`.

**5. Keluaran Berurutan Satu Baris:** Di dalam perulangan kedua, perintah `print(i, end=" ")` kembali mencetak nilai iterator secara horizontal dengan spasi. Pendekatan ini menunjukkan bagaimana parameter pertama pada fungsi `range()` dapat diatur untuk mengubah titik awal hitungan perulangan (mulai dari 1 alih-alih dari 0).

<h3><b>2.2.7 Program "perulangan_for3_2611532001.py"</b></h3>

```python
# Buat program untuk perulangan for dalam python
# Nama variabel ditambah 4 digit nim terakhir contoh: ulang_1234
# Program ini menggunakan fungsi input()

ulang_2001 = int(input("Masukkan jumlah perulangan: "))

jumlah_2001 = 0
for i_2001 in range(1, ulang_2001 + 1):
    print(i_2001, end=" ")
    jumlah_2001 = jumlah_2001 + i_2001

    if i_2001 < ulang_2001:
        print(" + ", end=" ")
    else:
        print(" = ", jumlah_2001, end=" ")
print()

print("Jumlah =", jumlah_2001)
```

**Penjelasan:**

**1. Menerima Masukan Jumlah Batas Deret:** Tanda `#` digunakan sebagai komentar penjelas. Sintaks `input()` menangkap masukan angka dari pengguna secara interaktif, yang kemudian dikonversi secara eksplisit menggunakan fungsi `int()` menjadi bilangan bulat pada variabel `ulang_2001`. Nilai ini menentukan batas akhir dari deret angka yang akan dijumlahkan.

**2. Inisialisasi Variabel Akumulator Penjumlahan:** Sebelum perulangan dimulai, variabel `jumlah_2001` dideklarasikan dan diinisialisasi dengan nilai awal 0. Variabel ini bertindak sebagai wadah penampung untuk mengakumulasikan total penjumlahan nilai angka secara bertahap pada setiap putaran iterasi.

**3. Pengendalian Perulangan for dan Pencetakan Angka:** Pernyataan `for i_2001 in range(1, ulang_2001 + 1):` mengontrol perulangan mulai dari angka 1 hingga batas `ulang_2001`. Di dalam blok perulangan, nilai iterator `i_2001` dicetak ke layar secara mendatar menggunakan argumen `end=" "` tanpa memindahkan kursor ke baris baru.

**4. Proses Akumulasi Penjumlahan Nilai:** Baris kode `jumlah_2001 = jumlah_2001 + i_2001` memperbarui nilai total penjumlahan secara progresif. Setiap kali perulangan berjalan, nilai iterator saat ini ditambahkan ke dalam variabel `jumlah_2001` sehingga nilainya terus bertambah naik.

**5. Penggunaan Percabangan untuk Pemformatan Simbol Matematika:** Program menggunakan struktur kondisional `if-else` di dalam perulangan untuk memeriksa apakah nilai iterator masih kurang dari batas akhir (`i_2001 < ulang_2001`). Jika ya, program mencetak simbol operator + sebagai penghubung deret. Namun, jika sudah mencapai iterasi terakhir, bagian `else` akan mengeksekusi pencetakan simbol = beserta total akumulasi sementara.

**6. Pemindahan Baris dan Pencetakan Hasil Akhir:** Perintah `print()` kosong dieksekusi setelah perulangan selesai untuk membuat baris baru. Selanjutnya, perintah `print("Jumlah =", jumlah_2001)` menampilkan informasi ringkasan hasil akhir total penjumlahan deret tersebut ke layar konsol.

```python
# --- HEADER DAN INPUT PROGRAM ---
print("=== PROGRAM JAM PASIR KRISTAL PALINDROMIK (PEKAN 5) ===")
# Meminta masukan skala N dari pengguna
n_2001 = int(input("Masukkan ukuran skala jam pasir (N): "))

# --- PERULANGAN UTAMA MATRIKS 2D ---
# Loop baris i (bergerak dari atas ke bawah)
for i_2001 in range(1, ((n_2001 * 2) + 4)):
    # Loop kolom j (bergerak dari kiri ke kanan)
    for j_2001 in range(1, ((n_2001 * 4) + 8)):

        # --- BINGKAI: PENGECEKAN 4 CORNER / SUDUT TERLUAR ---
        if ((i_2001 == 1 and j_2001 == 1) or (i_2001 == 1 and j_2001 == (n_2001 * 4) + 7) or (i_2001 == (n_2001 * 2) + 3 and j_2001 == (n_2001 * 4) + 7) or (i_2001 == (n_2001 * 2) + 3 and j_2001 == 1)):
            print("#", end="")

        else:
            # --- BINGKAI: ATAS, BAWAH, KIRI, DAN KANAN ---
            if (i_2001 == 1 or i_2001 == ((n_2001 * 2) + 3)):
               print("=", end="") # Garis horizontal pembatas atas dan bawah
            elif (j_2001 == 1 or j_2001 == ((n_2001 * 4) + 7)):
                print("|", end="") # Dinding vertikal pembatas kiri dan kanan

            else:
                # --- ELEMEN POROS KRISTAL TENGAH (<*>) ---
                if (j_2001 == ((n_2001 * 2) + 3)):
                    print("<", end="") # Karakter kiri poros
                elif (j_2001 == ((n_2001 * 2) + 5)):
                    print(">", end="") # Karakter kanan poros
                elif (j_2001 == ((n_2001 * 2) + 4)):
                    print("*", end="") # Karakter tengah poros
                elif (j_2001 == 2 or j_2001 == (n_2001 * 2) + 6):
                    print(" ", end="") # Spasi pemisah di dekat poros/bingkai

                else:
                    # --- MENENTUKAN SKALA ANGKA AKTIF PER BARIS ---
                    if i_2001 - 1 <= n_2001:
                        baris_2001 = (n_2001 - (i_2001 - 2)) # Fase Atas: Nilai N mengecil ke 1
                    else:
                        baris_2001 = (i_2001 - 2) - n_2001    # Fase Bawah: Nilai 1 membesar ke N

                    # --- PENCETAKAN DERET ANGKA PALINDROM ---
                    # Sisi Kiri Poros: Deret angka menurun (misal: 4 3 2 1)
                    if (j_2001 % 2 == 1) and (((n_2001 * 2) + 3 - (baris_2001 * 2)) <= j_2001 < ((n_2001 * 2) + 3)):
                        print(((n_2001 * 2) + 3 - j_2001) // 2, end="")
                    # Sisi Kanan Poros: Deret angka menaik (misal: 1 2 3 4)
                    elif (j_2001 % 2 == 1) and (((n_2001 * 2) + 5) < j_2001 <= ((n_2001 * 2) + 5 + (baris_2001 * 2))):
                        print((j_2001 - ((n_2001 * 2) + 5)) // 2, end="")
                    # Padding Spasi Kosong di Luar Jangkauan Angka
                    else:
                        print(" ", end="")

    # --- PINDAH KE BARIS BARU ---
    print()
```

Proses pembentukan pola menggunakan dua perulangan bersarang (_nested loop_). Di dalam perulangan kolom, struktur pengecekan pertama dilakukan untuk mencetak sudut luar matriks. Karakter pagar `#` dicetak secara khusus saat posisi baris dan kolom berada tepat di empat titik perpotongan terluar, yaitu sudut kiri-atas, kanan-atas, kiri-bawah, dan kanan-bawah.

**4. Pembentukan Dinding Pembatas dan Poros Kristal**

Jika posisi bukan merupakan sudut terluar, program mengecek kondisi untuk mencetak sisa bingkai dan poros tengah. Karakter garis tegak `|` dicetak pada seluruh baris di kolom pertama dan kolom terakhir sebagai dinding pembatas vertikal. Sebelum mencetak deret angka, program menentukan nilai variabel `baris_2001` secara dinamis untuk menentukan batas angka tertinggi pada baris tersebut. Sebaliknya, setelah melewati titik pusat menuju bagian bawah, nilainya bertambah kembali dari satu menuju skala N.

**6. Pencetakan Deret Angka Palindrom dan Perpindahan Baris**

Bagian akhir percabangan menangani pencetakan deret angka dengan syarat kolom ganjil `j_2001 % 2 == 1` agar angka terpisah oleh spasi secara otomatis. Pada sisi kiri poros, program mencetak deret angka menurun berdasarkan selisih jarak posisi kolom terhadap poros tengah. Pada sisi kanan poros, program mencetak deret angka menaik berdasarkan jarak posisi kolom dari batas kanan poros. Posisi lain di dalam bingkai yang tidak memenuhi syarat akan diisi oleh karakter spasi.
